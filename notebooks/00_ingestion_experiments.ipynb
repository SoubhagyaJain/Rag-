Use this notebook to:
- Inspect structure-aware chunks
- Compare different chunking parameters
- Verify rich metadata (page, section_path, has_code, has_table, etc.)

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent / "src"))

from rag.config import load_config
from rag.chunking.structure_aware import StructureAwareChunker
from rag.utils.io import read_json
```

```python
cfg = load_config("../config/default.yaml")
chunker = StructureAwareChunker(cfg)

result = chunker.chunk(cfg.paths.pdf)
print("Total chunks:", len(result.chunks))
print("Stats:", result.stats)
```

```python
# Inspect a few chunks with rich metadata
for ch in result.chunks[10:18]:
    print("="*80)
    print(ch.short_repr(160))
    print("Metadata:", {k: ch.metadata.get(k) for k in ["page", "chunk_type", "section_path", "has_table", "has_code"]})
    print()
```
